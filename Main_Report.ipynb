## 1. SUMMARY <a id='sec1'></a>
This report is on a 28-day forecast of daily pediatrics emergency attendances for an NHS acute trust hospital. It involves a time series data set of 1056 days, starting from March 2014 to February 2017. Results from this report is to support hospital administrators with informed decision making on medical staffing and effective patient management.

Exploratory data analysis showed that the year 2016 had the most emergencies, while 2014 and 2015 experienced the least number of cases. Also observed are day of week and month of year seasonality, with spikes observed in March, closely followed by November while August had the least daily attendances. Weekends had the most cases compared to weekdays.

Seasonal naive was chosen  as the naive benchmark and two prophet models were evaluated using rolling origin cross validation technique, forecast accuracy was measured by mean absolute error and mean absolute percentage error. The prophet model with the added holidays outperformed the basic model, this was selected as the main prophet model to be compared with the naive benchmark. Both models used a simple train-test split in the evaluation of prediction interval coverage(PIC).

Prophet was selected over seasonal naive based on its lower forecast errors and its better calibrated prediction intervals. The actual 28-day forecast, spanning from February  into March, shows a gradual increase in emergency department attendances with higher spikes observed during the weekends. However, short term forecast seems to be reliable for informed decision making, while a longer range of forecast horizon was associated with uncertainty.

## 2. Introduction <a id='sec2'></a>

An increasing demand for medical care places a huge importance on medical staffing and patient management. The Emergency department is the part of hospital that never goes to sleep and is operational all day and night, hence making it one of the busiest sections of a hospital, thus a crucial aspect of a hospital's administrative planning includes effective resource allocation and  staff scheduling. Appropriate staff management reduces worker's burnout and patient waiting times, which when poorly managed becomes chaotic for both patients and health workers given the need for urgency medical attention. Achieving this will require intelligent, reliable and reproducible forecasts on which administrators can make intelligible and informed decision.

This report addresses a task, set by an NHS Acute Trust to forecast the daily pediatrics emergency attendances over 28 days, the result of which is very vital in decisions pertaining to medical staffing and patient flow management. A time series data of approximately three years was provided, starting from April 2014 to February 2017. Forecasting time series techniques and models is to be applied to the data to identify a reliable and applicable method for an accurate short term forecast on daily pediatric emergency attendances.

#### 4.2 Forecasting methods <a id='sec6'></a>

Two prophet models were compared, a basic model and one including holidays against a naive benchmark. Prophet models were selected on account of the non-stationarity of the time series data due to trends and multiple seasonalities. 

##### Naive Benchmark
Due to the seasonality and trend of data, a seasonal naive model was selected as a benchmark. The naive model serves as a basic and simple forecast on which our complex model must necessarily outperform to be selected as a better and more accurate method. To evaluate for the absolute value of the forecast error, mean absolute error metrics (MAE) was used, because for scale independence and easy interpretability, mean absolute percentage error (MAPE) was employed. For better data representation and accuracy, rolling origin cross validation technique was used to estimate the model's point forecast errors. It consisted of a minimum training size of 916 days, a 28 day forward step which was used to prevent data leakage and a horizon of 7, 14, 21 and 28 days. 

A simple train-test split technique was used to assess prediction interval coverage by evaluating for empirical coverage, absolute coverage difference and winkler score. A test set of 28 days and a training set of 1028 days was used. A different split technique was employed for prediction interval estimation due to an absence of direct estimations on prediction interval using the inbuilt cross validation score under forecast tools library, Hence, for reliable and accurate comparison among models, a simple train test split technique was used. (Code and analysis seen on [Seasonal_naive](Seasonal_naive.ipynb) notebook)

##### Prophet 
The data was pre-processed into two columns for date (ds) and attendances (y) to fit into the prophet model. The models were created, the first model having only an interval width of 0.95 and to the second model was added United Kingdom holidays, this was to account for the attendances influenced by country holidays. Similar to the naive benchmark, estimation of point forecast accuracy was done using rolling origin cross validation  and evaluation of prediction interval was done using simple train test split with the same parameters for an unbiased comparison (Code and analysis seen at Section 1 on [Prophet_Model](PF.ipynb) notebook). A prophet model was selected based on lower point forecast errors.

##### Model Selection
The selected prophet model and naive benchmark were both compared and assessed based on the degree of point forecast error and prediction interval coverage. Following this, the most appropriate method was selected and an actual 28 day forecast on daily pediatrics attendances was done. (Code and analysis on the 28day forcast for NHS seen at Section 2 on [Prophet_model](PF.ipynb) notebook)

##### Seasonal Naive Model 
Naive method had an MAE of 10.69 and MAPE of 20.92% at 28days horizon. Over the 28days forecast,  a gradual decline in the magnitude of point forecast error was observed up to the third week of forecasting, however errors seemed to step back up at the 28th day. The model had an empirical coverage of hundred percent, making its coverage 5 percent higher than the nominal coverage (ACD = 0.05). Winkler score method was used to inspect the model coverage to ensure its sharper and well calibrated, models with lower scores are preferred. Seasonal naive attained a score of 64.6. ## 6. Discussion <a id='sec11'></a>
The time series data explored was a non-stationary data with trend and multiple seasonalities. For better data representation, Cross validation technique was used to evaluate errors and for a fair comparison, a simple train test split was used to assess coverage. Prophet stood out when compared to the naive bench mark based on its lower forecast errors and its better prediction interval coverage. The hospital is observed to experience busy days on weekends with March having the most activity and  August having the least engagements. The 28 day forecast into march closely aligns with the hospital's data exploratory analysis findings, however forecast uncertainty is observed as the horizon increases.

##### Recommendations
To prevent medical staff burn out and handle medical demands, there should be a re-assessment and re-adjusted of staffing to match demand. Weekends shifts should be flexible and not protracted over long hours, to further prevent fatigue and exhaustion, there should be allocation of more staff to weekend duties. Prompt referral of severe cases requiring specialist care and effective triage processes in the emergency department should help in reducing over-crowding

##### Limitations
Although the model included holidays as an external influence, it falls short of other potential external predictors like weather forecast and seasonal illnesses commonly observed in pediatrics group. Only two forecasting methods were evaluated, alternative modeling approaches such as deep learning and other statistical models that could potentially provide a better forecast performance were not explored. Further evaluation comparing a wide range of models on daily pediatrics emergency department attendance is warranted.


## 6. Conclusion <a id='sec12'></a>
The prophet model had the best performance as it captured trends and seasonality within the data and factored in country holidays which strongly influenced the daily pediatrics emergency attendances. it achieved better coverage and lower errors compared to our naive bench. Uncertainty increases over longer horizons, so interpretation should be taken cautiously.
